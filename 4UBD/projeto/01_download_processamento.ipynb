```python
import os
import geopandas as gpd
import pandas as pd
from shapely.geometry import Point

# cria pasta
os.makedirs("dados", exist_ok=True)

# dados fictícios
dados = {
    'municipio': [
        'Uberlandia',
        'Belo Horizonte',
        'Montes Claros'
    ],
    'bioma': [
        'Cerrado',
        'Mata Atlantica',
        'Cerrado'
    ],
    'data': [
        '2022-08-10',
        '2022-09-15',
        '2022-08-22'
    ],
    'area_ha': [
        120,
        45,
        300
    ],
    'longitude': [
        -48.2772,
        -43.9386,
        -43.865
    ],
    'latitude': [
        -18.9186,
        -19.9167,
        -16.7286
    ]
}

# dataframe
df = pd.DataFrame(dados)

# cria geometria
geometry = gpd.points_from_xy(
    df.longitude,
    df.latitude
)

# geodataframe
gdf = gpd.GeoDataFrame(
    df,
    geometry=geometry,
    crs='EPSG:4326'
)

# reprojeta
gdf = gdf.to_crs(epsg=31983)

# salva geojson
gdf.to_file(
    'dados/focos-desmatamento-mg.geojson',
    driver='GeoJSON'
)

print(gdf.head())
```
