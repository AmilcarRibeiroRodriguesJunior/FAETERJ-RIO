```python
!pip install geopandas -q

import geopandas as gpd
import os

url_do_arquivo = (
    'https://raw.githubusercontent.com/'
    'tbrugz/geodata-br/master/geojson/'
    'geojs-31-mun.json'
)

diretorio = 'dados'

caminho_arquivo = (
    f'{diretorio}/municipios-mg.geojson'
)


os.makedirs(
    diretorio,
    exist_ok=True
)

print("Baixando municípios de MG...")

gdf = gpd.read_file(
    url_do_arquivo
)

print(
    f'Total de municípios: {len(gdf)}'
)

print(
    'Reprojetando para EPSG:31983...'
)

gdf = gdf.to_crs(
    epsg=31983
)

print(
    'Calculando área em km²...'
)

gdf['area_km2'] = (
    gdf.geometry.area / 1_000_000
)

print(
    f'Salvando arquivo em: {caminho_arquivo}'
)

gdf.to_file(
    caminho_arquivo,
    driver='GeoJSON'
)

print(
    'Processo concluído com sucesso!'
)

print(gdf.head())
```

## Interpretação da Correlação entre PIB e Desmatamento

Observa-se uma tendência de municípios com maior atividade econômica apresentarem também maiores áreas desmatadas. Esse comportamento pode estar associado à expansão agropecuária e ao avanço de atividades econômicas sobre áreas naturais, especialmente no bioma Cerrado.

Além disso, municípios com maior PIB tendem a possuir maior infraestrutura produtiva, o que pode contribuir para a pressão sobre recursos naturais e expansão territorial.

Os resultados indicam a necessidade de intensificar ações de fiscalização ambiental em municípios economicamente estratégicos, principalmente aqueles localizados em regiões com maior incidência de desmatamento.
